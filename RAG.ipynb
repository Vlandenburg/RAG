```python
%pip install rich
from rich.console import Console

console = Console()
```

```python
%pip install datasets
from datasets import load_dataset

dataset = load_dataset('AiresPucrs/tmdb-5000-movies', split='train')

console.print(dataset)
```

```python
%pip install qdrant-client sentence-transformers
from qdrant_client import models, QdrantClient
from sentence_transformers import SentenceTransformer

qdrant = QdrantClient(":memory:")
encoder = SentenceTransformer('sentence-transformers/all-mpnet-base-v2')

console.print(encoder)
```

```python
collection_name = "movies"


qdrant.recreate_collection(
    collection_name=collection_name,
    vectors_config=models.VectorParams(
        size=encoder.get_sentence_embedding_dimension(),
        distance=models.Distance.COSINE,
    ),
)
```

```python
from qdrant_client.models import PointStruct
from tqdm.auto import tqdm

points = []

# Идем по строкам датасета
for row in tqdm(dataset):
    if not row["overview"]:
        continue
    vector = encoder.encode(row["overview"]).tolist()

    points.append(PointStruct(id=row["id"], vector=vector, payload=dict(row)))

qdrant.upload_points(collection_name=collection_name, points=points)
```

```python
console.print(qdrant.get_collection(collection_name=collection_name))
```

```python
user_prompt = "Love story between an Asian king and European teacher"

# Кодируем запрос тем же энкодером в 768 чисел
query_vector = encoder.encode(user_prompt).tolist()
```

```python
from qdrant_client import models

query_filter = models.Filter(
    must=[
        models.FieldCondition(
            key="release_date",
            range=models.DatetimeRange(gte="1990-01-01", lte="1999-12-31"),
        )
    ]
)
```

```python
response = qdrant.query_points(
    collection_name=collection_name,
    query=query_vector,
    query_filter=query_filter,
    limit=1,
)

hits = response.points
```

```python
from rich.table import Table

table = Table(title="Retrieval Results", show_lines=True)

table.add_column("ID", style="#e0e0e0")
table.add_column("Original Title", style="#e0e0e0")
table.add_column("Overview", style="bright_red")
table.add_column("Score", style="#89ddff")

for hit in hits:
    table.add_row(
        str(hit.payload["id"]),
        hit.payload["original_title"],
        f'{hit.payload["overview"]}',
        f"{hit.score:.4f}",
    )

console.print(table)
```

```python
fields = ["original_title", "title", "overview", "release_date", "popularity"]

search_results = [
    {
        field: hit.payload.get(field)
        for field in fields
        if hit.payload.get(field) is not None
    }
    for hit in hits
]

console.print(search_results)
```

```python
from rich.panel import Panel
from rich.text import Text
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct",
    device="cpu",
)

context_str = f"Found movies: {search_results}"

messages = [
    {
        "role": "system",
        "content": "You are a movie expert. Recommend a movie to the user based on the provided search results. Be concise and friendly.",
    },
    {
        "role": "user",
        "content": f"User query: {user_prompt}\n\nContext:\n{context_str}",
    },
]

output = generator(messages, max_new_tokens=150, temperature=0.7)

response_content = output[0]["generated_text"][-1]["content"]

styled_panel = Panel(
    Text(response_content),
    title="Movie Recommendation with Retrieval (Local Qwen)",
    expand=False,
    border_style="bright_yellow",
    padding=(1, 1),
)

console.print(styled_panel)
```
